```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data")

dataset_files = [
    "CEAS_08.csv",
    "Enron.csv",
    "Ling.csv",
    "Nazario.csv",
    "Nigerian_Fraud.csv",
    "phishing_email.csv",
    "SpamAssasin.csv"
]

datasets = {}

for filename in dataset_files:
    path = DATA_DIR / filename
    
    try:
        df = pd.read_csv(path)
        datasets[filename] = df
        
        print("=" * 80)
        print(f"DATASET: {filename}")
        print(f"Shape: {df.shape}")
        print(f"Columns: {df.columns.tolist()}")
        print()
        print(df.head(2))
        print()
        
    except Exception as e:
        print(f"ERROR loading {filename}: {e}")
```

```python
#inspecting the labels
for name, df in datasets.items():
    print("=" * 80)
    print(f"{name}")
    
    print("\nColumns:")
    print(df.columns.tolist())
    
    print("\nPossible label columns:")
    for col in df.columns:
        unique_count = df[col].nunique(dropna=False)
        
        if unique_count <= 10:
            print(f"  {col}: {unique_count} unique values")
            print(f"     {df[col].value_counts(dropna=False).to_dict()}")
```

```python
#Investigating the label semantics
for name, df in datasets.items():
    print("=" * 80)
    print(name)
    
    if "label" in df.columns:
        for label in sorted(df["label"].dropna().unique()):
            sample = df[df["label"] == label]
            
            print(f"\nLABEL: {label}")
            print(f"Number of samples: {len(sample)}")
            
            cols = [c for c in ["subject", "body", "text_combined"] if c in df.columns]
            
            if cols:
                display(sample[cols].head(3))
```

```python
#checking the overlap and duplicacy in the dataset
for name, df in datasets.items():
    if "body" in df.columns:
        print(name, "body duplicates:", df["body"].duplicated().sum())

    if "text_combined" in df.columns:
        print(name, "text duplicates:", df["text_combined"].duplicated().sum())
```

```python
# Create normalized text for overlap checking

def normalize_text(text):
    if pd.isna(text):
        return ""
    
    text = str(text).lower()
    text = " ".join(text.split())
    return text


# Create normalized body/text for each dataset
normalized_texts = {}

for name, df in datasets.items():
    
    if "text_combined" in df.columns:
        normalized_texts[name] = set(
            df["text_combined"].dropna().map(normalize_text)
        )
        
    elif "body" in df.columns:
        normalized_texts[name] = set(
            df["body"].dropna().map(normalize_text)
        )


# Compare phishing_email.csv against every other dataset
phishing_texts = normalized_texts["phishing_email.csv"]

print("phishing_email unique texts:", len(phishing_texts))
print()

for name, texts in normalized_texts.items():
    
    if name == "phishing_email.csv":
        continue
    
    overlap = phishing_texts.intersection(texts)
    
    print(f"{name}: {len(overlap):,} overlapping messages")
```

```python
#inspecting the actual label distribution and characteristics of phishing_email.csv
df = datasets["phishing_email.csv"]

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nLabel distribution:")
print(df["label"].value_counts())

print("\nLabel percentages:")
print(df["label"].value_counts(normalize=True) * 100)

print("\nSample messages by label:")

for label in sorted(df["label"].unique()):
    print(f"\n===== LABEL {label} =====")
    samples = df[df["label"] == label]["text_combined"].dropna().sample(
        min(10, len(df[df["label"] == label])),
        random_state=42
    )

    for i, text in enumerate(samples, 1):
        print(f"\n--- Example {i} ---")
        print(str(text)[:500])
```

```python
#dataset size and characteristics
for name, df in datasets.items():

    if "text_combined" in df.columns:
        text_col = "text_combined"
    else:
        text_col = "body"

    text_lengths = df[text_col].fillna("").astype(str).str.len()

    print("\n" + "=" * 70)
    print(name)

    print("Average text length :", round(text_lengths.mean(), 2))
    print("Median text length  :", round(text_lengths.median(), 2))
    print("Minimum text length :", text_lengths.min())
    print("Maximum text length :", text_lengths.max())

    print("Empty messages      :", (text_lengths == 0).sum())
```
