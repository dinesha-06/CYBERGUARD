```python
#all imports
import pandas as pd
import numpy as np
from pathlib import Path

DATA_DIR = Path(r"C:\Users\HP\OneDrive\Desktop\BPUT\CYBERGUARD\malicious_url\data\raw")

print("Dataset directory:", DATA_DIR.resolve())
print("Files found:")

for file in DATA_DIR.iterdir():
    print(file.name)
```

```python
#loading both datasets
phiusiil_path = DATA_DIR / "PhiUSIIL_Phishing_URL_Dataset.csv"
uci_path = DATA_DIR / "uci-ml-phishing-dataset.csv"

phiusiil_df = pd.read_csv(phiusiil_path)
uci_df = pd.read_csv(uci_path)

print("PhiUSIIL shape:", phiusiil_df.shape)
print("UCI shape:", uci_df.shape)
```

```python
#inspaecting the scheme
print("========== PHIUSIIL ==========")
print(phiusiil_df.columns.tolist())
display(phiusiil_df.head())

print("\n========== UCI ==========")
print(uci_df.columns.tolist())
display(uci_df.head())
```

```python
#inspecting the target labels 
print("PhiUSIIL label distribution:")
print(phiusiil_df["label"].value_counts(dropna=False))

print("\nUCI target candidates:")
for col in ["Result", "result", "label", "Label", "class"]:
    if col in uci_df.columns:
        print(f"\n{col}:")
        print(uci_df[col].value_counts(dropna=False))
```

```python
#inspecting the missing or dupllicates value
def inspect_dataset(df, name):
    print(f"\n===== {name} =====")
    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Duplicate rows:", df.duplicated().sum())
    print("Columns with missing values:")
    print(df.isna().sum()[df.isna().sum() > 0])

inspect_dataset(phiusiil_df, "PhiUSIIL")
inspect_dataset(uci_df, "UCI")
```

```python
#inspecting the features type
print("PhiUSIIL data types:")
display(phiusiil_df.dtypes.value_counts())

print("\nUCI data types:")
display(uci_df.dtypes.value_counts())
```
